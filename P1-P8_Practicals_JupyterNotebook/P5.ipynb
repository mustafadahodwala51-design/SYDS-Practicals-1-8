### 1. Create a Pandas DataFrame

**Question:** Create a Pandas DataFrame containing student names and marks.

````python
import pandas as pd
data = {
    "Name": ["Rahul", "Priya", "Amit", "Sneha"],
    "Marks": [75, 85, 65, 90]
}
df = pd.DataFrame(data)
print(df)
````

**Sample Output:**
```
     Name  Marks
0   Rahul     75
1   Priya     85
2    Amit     65
3   Sneha     90
```

### 2. Display Statistical Information

**Question:** Create a DataFrame of student marks and display statistical information.

````python
import pandas as pd
data = {
    "Name": ["Rahul", "Priya", "Amit", "Sneha"],
    "Marks": [75, 85, 65, 90]
}
df = pd.DataFrame(data)
print(df)
print()
print("Statistical Information:")
print(df["Marks"].describe())
````

### 3. Create Pandas Series from Dictionary

**Question:** Create a Pandas Series using a dictionary containing student names and marks.

````python
import pandas as pd
marks = {
    "Rahul": 75,
    "Priya": 85,
    "Amit": 65,
    "Sneha": 90
}
series = pd.Series(marks)
print(series)
````

### 4. Filter Pandas Series

**Question:** Create a Pandas Series of marks and display only marks greater than 70.

````python
import pandas as pd
marks = pd.Series([55, 75, 80, 60, 90])
result = marks[marks > 70]
print("Marks greater than 70:")
print(result)
````

### 5. Simple Student Data Analysis

**Question:** Create a DataFrame containing student names, marks and attendance. Display students who have marks greater than 70.

````python
import pandas as pd
data = {
    "Name": ["Rahul", "Priya", "Amit", "Sneha", "Kiran"],
    "Marks": [75, 85, 60, 90, 65],
    "Attendance": [80, 90, 75, 95, 70]
}
df = pd.DataFrame(data)
print("Student Data:")
print(df)
print()
print("Students scoring more than 70:")
print(df[df["Marks"] > 70])
````
