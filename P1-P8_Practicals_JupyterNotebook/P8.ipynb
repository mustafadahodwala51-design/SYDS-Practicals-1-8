### 1. Recursion - Factorial

**Question:** Write a program to find the factorial of a number using recursion.

````python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

num = int(input("Enter number: "))
print("Factorial:", factorial(num))
````

**Sample Output:**
```
Enter number: 5
Factorial: 120
```

### 2. Recursion - Fibonacci Series

**Question:** Write a program to generate the Fibonacci series using recursion.

````python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

n = int(input("Enter number of terms: "))
for i in range(n):
    print(fibonacci(i), end=" ")
````

**Sample Output:**
```
Enter number of terms: 8
0 1 1 2 3 5 8 13
```

### 3. Binary Tree and Inorder Traversal

**Question:** Create a simple binary tree and perform inorder traversal.

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

def inorder(root):
    if root:
        inorder(root.left)
        print(root.data)
        inorder(root.right)

root = Node(10)
root.left = Node(20)
root.right = Node(30)

print("Root:", root.data)
print("Left:", root.left.data)
print("Right:", root.right.data)
print()
print("Inorder Traversal:")
inorder(root)
````

**Sample Output:**
```
Root: 10
Left: 20
Right: 30

Inorder Traversal:
20
10
30
```

### 4. Dictionary and Frequency of Elements

**Question:** Write a program using a dictionary to count the frequency of elements and words.

````python
numbers = [10, 20, 10, 30, 20, 10]
frequency = {}
for number in numbers:
    if number in frequency:
        frequency[number] += 1
    else:
        frequency[number] = 1
print("Element frequency:", frequency)

text = "python data science python data"
words = text.split()
word_frequency = {}
for word in words:
    if word in word_frequency:
        word_frequency[word] += 1
    else:
        word_frequency[word] = 1
print("Word frequency:", word_frequency)
````

**Sample Output:**
```
Element frequency: {10: 3, 20: 2, 30: 1}
Word frequency: {'python': 2, 'data': 2, 'science': 1}
```
