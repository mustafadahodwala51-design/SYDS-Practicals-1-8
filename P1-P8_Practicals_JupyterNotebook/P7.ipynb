### 1. Implement Stack Using List

**Question:** Write a program to implement a stack using a list (LIFO - Last In First Out).

````python
stack = []
stack.append(10)
stack.append(20)
stack.append(30)
print("Stack:", stack)
print("Deleted:", stack.pop())
print("Stack after deletion:", stack)
````

**Sample Output:**
```
Stack: [10, 20, 30]
Deleted: 30
Stack after deletion: [10, 20]
```

### 2. Stack Using Menu

**Question:** Write a menu-driven program to perform stack operations (push, pop, display).

````python
stack = []
while True:
    print()
    print("1. Push")
    print("2. Pop")
    print("3. Display")
    print("4. Exit")
    choice = int(input("Enter choice: "))
    if choice == 1:
        value = int(input("Enter value: "))
        stack.append(value)
    elif choice == 2:
        if len(stack) == 0:
            print("Stack is empty")
        else:
            print("Deleted:", stack.pop())
    elif choice == 3:
        print("Stack:", stack)
    elif choice == 4:
        break
    else:
        print("Invalid choice")
````

### 3. Implement Queue Using List

**Question:** Write a program to implement a queue using a list (FIFO - First In First Out).

````python
queue = []
queue.append(10)
queue.append(20)
queue.append(30)
print("Queue:", queue)
print("Deleted:", queue.pop(0))
print("Queue after deletion:", queue)
````

**Sample Output:**
```
Queue: [10, 20, 30]
Deleted: 10
Queue after deletion: [20, 30]
```

### 4. Queue Using Menu

**Question:** Write a menu-driven program to perform queue operations (insert, delete, display).

````python
queue = []
while True:
    print()
    print("1. Insert")
    print("2. Delete")
    print("3. Display")
    print("4. Exit")
    choice = int(input("Enter choice: "))
    if choice == 1:
        value = int(input("Enter value: "))
        queue.append(value)
    elif choice == 2:
        if len(queue) == 0:
            print("Queue is empty")
        else:
            print("Deleted:", queue.pop(0))
    elif choice == 3:
        print("Queue:", queue)
    elif choice == 4:
        break
    else:
        print("Invalid choice")
````

### 5. Circular Queue

**Question:** Write a program to implement a circular queue using deque.

````python
from collections import deque
queue = deque(maxlen=3)
queue.append(10)
queue.append(20)
queue.append(30)
print(queue)
queue.append(40)
print(queue)
````

**Sample Output:**
```
deque([10, 20, 30], maxlen=3)
deque([20, 30, 40], maxlen=3)
```
